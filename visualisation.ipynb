```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generowanie danych
x = np.linspace(-10, 10, 100)  # 100 punktów od -10 do 10
y = 1 / (1 + np.exp(-x)) * 2 - 1
#y = np.fabs(x) ** 0.5
#y = x ** 0.5

# Tworzenie wykresu
sns.set_theme(style="whitegrid")  # Ustawienie stylu Seaborn
sns.lineplot(x=x, y=y)

# Wyświetlenie wykresu
plt.xlabel("x")
plt.ylabel("y")
plt.title("Wykres funkcji sigmoidalnej")
plt.show()
```
