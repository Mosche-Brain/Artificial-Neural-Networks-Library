```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generowanie danych
x = np.linspace(-10, 10, 100)
sigmoid = 1 / (1 + np.exp(-x))
y = sigmoid * 2 - 1  # funkcja aktywacji: zeskalowana sigmoid
y_prim = y * (1 - y)  # pochodna zeskalowanej sigmoidy

# Tworzenie wykresu
sns.set_theme(style="whitegrid")

# Rysowanie dwóch linii
sns.lineplot(x=x, y=y, label="f(x) = 2 * sigmoid(x) - 1")
sns.lineplot(x=x, y=y_prim, label="f'(x)")

# Etykiety i tytuł
plt.xlabel("x")
plt.ylabel("y")
plt.title("Funkcja i jej pochodna")
plt.legend()  # dodanie legendy

# Wyświetlenie wykresu
plt.show()
```
